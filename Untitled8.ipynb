```python
# Step 1: Import required libraries
import nltk
from nltk.tokenize import word_tokenize
from gensim.models import Word2Vec

# Step 2: Download tokenizer data
nltk.download('punkt')
nltk.download('punkt_tab') # Added to download the missing resource

# Step 3: Define corpus
corpus = [
    "machine learning is evolving",
    "AI is changing"
]

# Step 4: Tokenize the corpus
tokenized_corpus = [word_tokenize(sentence.lower()) for sentence in corpus]

print("Tokenized Corpus:")
for i, tokens in enumerate(tokenized_corpus):
    print(f"Sentence {i+1}: {tokens}")

# Step 5: Train Word2Vec model (Embeddings)
model = Word2Vec(
    sentences=tokenized_corpus,
    vector_size=50,   # size of embedding vector
    window=2,         # context window
    min_count=1,      # include all words
    workers=1,
    sg=0              # CBOW model (use sg=1 for Skip-gram)
)

# Step 6: Get embeddings for words
print("\nWord Embeddings:\n")
for word in model.wv.index_to_key:
    print(f"{word}: {model.wv[word]}")
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# Load a pre-trained Hugging Face model
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')

# Input texts
text1 = "Artificial intelligence is changing the world"
text2 = "AI is transforming how the world works"

# Convert texts into embeddings (numerical vectors)
embedding1 = model.encode([text1])
embedding2 = model.encode([text2])

# Compute cosine similarity
similarity_score = cosine_similarity(embedding1, embedding2)

print("Similarity Score:", similarity_score[0][0])
```

```python
from transformers import pipeline

sentiment_analyzer = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

text = "I really enjoyed this movie. The acting was amazing!"
result = sentiment_analyzer(text)

print(result)
```

```python
from transformers import pipeline

# Load a fake news classification model from Hugging Face
fake_news_classifier = pipeline(
    "text-classification",
    model="hamzab/roberta-fake-news-classification"
)

# Input news text
news_text = """
The government has announced a new policy that will provide free electricity to all citizens starting next month.
"""

# Classify the news
result = fake_news_classifier(news_text)

# Print result
print(result)
```
