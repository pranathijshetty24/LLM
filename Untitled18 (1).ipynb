### Named Entity Recognition with a Custom Dataset

This example demonstrates Named Entity Recognition (NER) using a custom, small dataset. It defines a simple vocabulary, sample sentences with corresponding entity tags, and then trains a Bidirectional LSTM model to predict these tags.

ENTITY RECOGNITION

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Bidirectional, LSTM, Dense, TimeDistributed
from tensorflow.keras.preprocessing.sequence import pad_sequences

# 1. Define a small vocabulary and entity tags
vocab = {
    'PAD': 0, 'UNK': 1, 'the': 2, 'quick': 3, 'brown': 4, 'fox': 5, 'jumps': 6, 'over': 7,
    'lazy': 8, 'dog': 9, 'New': 10, 'York': 11, 'is': 12, 'a': 13, 'city': 14, 'Google': 15,
    'is': 16, 'a': 17, 'company': 18, 'I': 19, 'love': 20, 'Paris': 21, 'and': 22, 'London': 23,
    'He': 24, 'went': 25, 'to': 26, 'Germany': 27, 'today': 28
}

tag_to_index = {'O': 0, 'B-LOC': 1, 'I-LOC': 2, 'B-ORG': 3, 'I-ORG': 4, 'B-PER': 5, 'I-PER': 6}
index_to_tag = {i: tag for tag, i in tag_to_index.items()}

# 2. Create sample sentences and corresponding NER tags
sentences = [
    ['New', 'York', 'is', 'a', 'city'],
    ['Google', 'is', 'a', 'company'],
    ['I', 'love', 'Paris', 'and', 'London'],
    ['He', 'went', 'to', 'Germany', 'today']
]

tags = [
    ['B-LOC', 'I-LOC', 'O', 'O', 'O'],
    ['B-ORG', 'O', 'O', 'O', 'O'],
    ['O', 'O', 'B-LOC', 'O', 'B-LOC'],
    ['O', 'O', 'O', 'B-LOC', 'O']
]

# 3. Convert words and tags to numerical sequences
X = [[vocab.get(word, vocab['UNK']) for word in s] for s in sentences]
y = [[tag_to_index[t] for t in ts] for ts in tags]

# 4. Pad sequences to ensure uniform length
max_len = 10
X = pad_sequences(X, maxlen=max_len, padding='post', value=vocab['PAD'])
y = pad_sequences(y, maxlen=max_len, padding='post', value=tag_to_index['O'])

# Reshape y for TimeDistributed layer
y = np.expand_dims(y, -1)

# 5. Build a simple Keras NER model
# Calculate the correct input_dim for the Embedding layer
embedding_input_dim = max(vocab.values()) + 1

model = Sequential([
    Embedding(embedding_input_dim, 50, input_length=max_len),
    Bidirectional(LSTM(100, return_sequences=True)),
    TimeDistributed(Dense(len(tag_to_index), activation='softmax'))
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("Model Summary:")
model.summary()

# 6. Train the model (using a very small dataset for demonstration)
print("\nTraining Model...")
model.fit(X, y, epochs=10, batch_size=1, verbose=0)
print("Training Complete.")

# 7. Make a prediction on a new sample
new_sentence = ['I', 'love', 'New', 'York', 'and', 'Google']
new_sentence_numerical = [vocab.get(word, vocab['UNK']) for word in new_sentence]
new_sentence_padded = pad_sequences([new_sentence_numerical], maxlen=max_len, padding='post', value=vocab['PAD'])

prediction = model.predict(new_sentence_padded)
predicted_tags_indices = np.argmax(prediction, axis=-1)

print("\nNamed Entity Recognition Output for new sentence:")
# Create index_to_word mapping for clear output
index_to_word = {v: k for k, v in vocab.items()}

for i, word_idx in enumerate(new_sentence_numerical):
    if i < max_len: # Ensure we don't go past max_len if sentence is longer
        word = index_to_word.get(word_idx, 'UNK') # Use .get with default for safety
        predicted_tag = index_to_tag[predicted_tags_indices[0][i]]
        print(f"Word: '{word}' -> Predicted Tag: {predicted_tag}")
```

### Sentiment Analysis of Movie Reviews

This example demonstrates how to build a simple sentiment analysis model to classify movie reviews as positive or negative. It uses the IMDb movie review dataset and a basic Feedforward Neural Network.

```python
import tensorflow as tf
from tensorflow.keras.layers import Embedding, Flatten, Dense
from tensorflow.keras.models import Sequential
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np

# 1. Load the IMDb dataset
vocab_size = 10000  # Consider the top 10,000 most frequent words
max_review_len = 256 # Max length of reviews

(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=vocab_size)

# Pad sequences to ensure uniform length for all reviews
X_train = pad_sequences(X_train, maxlen=max_review_len)
X_test = pad_sequences(X_test, maxlen=max_review_len)

print(f"X_train shape: {X_train.shape}")
print(f"y_train shape: {y_train.shape}")

# 2. Build a simple sentiment classification model
model_sentiment = Sequential([
    Embedding(vocab_size, 16, input_length=max_review_len),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(1, activation='sigmoid') # Sigmoid for binary classification
])

model_sentiment.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

print("\nSentiment Model Summary:")
model_sentiment.summary()

# 3. Train the model
print("\nTraining Sentiment Model...")
history = model_sentiment.fit(
    X_train,
    y_train,
    epochs=5, # Reduced epochs for quicker demonstration
    batch_size=32,
    validation_data=(X_test, y_test),
    verbose=0
)
print("Sentiment Model Training Complete.")

# 4. Evaluate the model
loss, accuracy = model_sentiment.evaluate(X_test, y_test, verbose=0)
print(f"\nTest Accuracy: {accuracy*100:.2f}%")

# 5. Make a prediction on a new review
word_index = imdb.get_word_index()
reverse_word_index = {v: k for k, v in word_index.items()}

def decode_review(text):
    return ' '.join([reverse_word_index.get(i - 3, '?') for i in text])

# Get a sample test review
sample_review_text = decode_review(X_test[0])
print(f"\nSample Review: {sample_review_text[:200]}...")

# Preprocess the sample review for prediction
sample_review_padded = X_test[0].reshape(1, max_review_len)

# Predict sentiment
prediction_sentiment = model_sentiment.predict(sample_review_padded)

# Interpret the prediction
sentiment = "Positive" if prediction_sentiment[0][0] >= 0.5 else "Negative"
print(f"Predicted Sentiment: {sentiment} (Confidence: {prediction_sentiment[0][0]*100:.2f}%)")
```

Question Answering

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np

# -----------------------------
# Load IMDB Dataset
# -----------------------------
vocab_size = 1000
max_len = 20

(x_train, y_train), (x_test, y_test) = keras.datasets.imdb.load_data(
    num_words=vocab_size
)

x_train = keras.preprocessing.sequence.pad_sequences(x_train, maxlen=max_len)
x_test = keras.preprocessing.sequence.pad_sequences(x_test, maxlen=max_len)

# -----------------------------
# Transformer Encoder
# -----------------------------
class TransformerEncoder(layers.Layer):
    def __init__(self, embed_dim, num_heads, ff_dim):
        super().__init__()
        self.att = layers.MultiHeadAttention(num_heads=num_heads, key_dim=embed_dim)
        self.ffn = keras.Sequential([
            layers.Dense(ff_dim, activation="relu"),
            layers.Dense(embed_dim)
        ])
        self.norm1 = layers.LayerNormalization()
        self.norm2 = layers.LayerNormalization()

    def call(self, x):
        attn_out = self.att(x, x)
        x = self.norm1(x + attn_out)
        ffn_out = self.ffn(x)
        return self.norm2(x + ffn_out)

# -----------------------------
# Model Definition
# -----------------------------
embed_dim = 128
num_heads = 4
ff_dim = 256

inputs = layers.Input(shape=(max_len,))
embedding = layers.Embedding(vocab_size, embed_dim)(inputs)
encoder = TransformerEncoder(embed_dim, num_heads, ff_dim)(embedding)

x = layers.GlobalAveragePooling1D()(encoder)
x = layers.Dense(64, activation="relu")(x)
output = layers.Dense(1, activation="sigmoid")(x)

model = keras.Model(inputs, output)

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# -----------------------------
# Train Model
# -----------------------------
model.fit(
    x_train,
    y_train,
    epochs=3,
    batch_size=64,
    validation_data=(x_test, y_test)
)

# -----------------------------
# YES / NO Prediction
# -----------------------------
sample_review = x_test[0].reshape(1, max_len)

prediction = model.predict(sample_review)[0][0]

if prediction >= 0.5:
    print("Answer: YES (Positive Review)")
else:
    print("Answer: NO (Negative Review)")

print("Confidence:", round(float(prediction), 2))
```

Text Summarization


```python
#text summarization
from tensorflow.keras.datasets import imdb
from collections import Counter

# Load IMDb dataset
(X_train, _), _ = imdb.load_data(num_words=1000)

# Get word index
word_index = imdb.get_word_index()

# Reverse mapping with offset handling
reverse_word_index = {value + 3: key for key, value in word_index.items()}
reverse_word_index[0] = "<PAD>"
reverse_word_index[1] = "<START>"
reverse_word_index[2] = "<UNK>"

# Decode review function
def decode_review(review):
    return " ".join([reverse_word_index.get(i, "?") for i in review])

# -------- USER INPUT --------
review_no = int(input("Enter IMDb review number (0–24999): "))

review_text = decode_review(X_train[review_no])
words = review_text.split()

# Frequency-based extractive summarization
freq = Counter(words)
summary = " ".join([word for word, _ in freq.most_common(15)])

# Output
print("\nSelected IMDb Review (partial):\n")
print(review_text[:300], "...")

print("\nSummary:\n")
print(summary)
```
