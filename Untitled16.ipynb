```python
# Simple encoder for text summarization

def encoder(text):
    stopwords = {
        "is", "a", "the", "an", "here", "this", "that",
        "and", "or", "to", "of", "in", "on"
    }

    words = text.lower().split()

    encoded_output = []
    for word in words:
        if word not in stopwords:
            encoded_output.append(word)

    return encoded_output


# Test input
sentence = "apple is a healthy fruit here"
print("Encoder Output:", encoder(sentence))
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# -------------------- DATA --------------------
input_sentence = "apple is a healthy fruit here"
summary_sentence = "apple healthy fruit"

def tokenize(text):
    return text.lower().split()

vocab = {"<pad>": 0, "<sos>": 1, "<eos>": 2}
for word in tokenize(input_sentence + " " + summary_sentence):
    if word not in vocab:
        vocab[word] = len(vocab)

inv_vocab = {v: k for k, v in vocab.items()}

def encode(text):
    return [vocab[w] for w in tokenize(text)]

src = [vocab["<sos>"]] + encode(input_sentence) + [vocab["<eos>"]]
trg = [vocab["<sos>"]] + encode(summary_sentence) + [vocab["<eos>"]]

src_tensor = torch.tensor(src).unsqueeze(1)
trg_tensor = torch.tensor(trg)

# -------------------- MODEL --------------------
class Encoder(nn.Module):
    def __init__(self, vocab_size, embed_size, hidden_size):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_size)
        self.lstm = nn.LSTM(embed_size, hidden_size)

    def forward(self, x):
        emb = self.embedding(x)
        _, (hidden, cell) = self.lstm(emb)
        return hidden, cell

class Decoder(nn.Module):
    def __init__(self, vocab_size, embed_size, hidden_size):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_size)
        self.lstm = nn.LSTM(embed_size, hidden_size)
        self.fc = nn.Linear(hidden_size, vocab_size)

    def forward(self, x, hidden, cell):
        # x is a scalar tensor (0-D tensor) representing a single token ID
        # self.embedding(x) will produce a (embed_size,) tensor
        # To make it (seq_len, batch_size, input_size) -> (1, 1, embed_size) for LSTM
        emb = self.embedding(x).unsqueeze(0).unsqueeze(0)

        out_lstm, (hidden, cell) = self.lstm(emb, (hidden, cell))
        # out_lstm is (1, 1, hidden_size)
        # Apply FC layer, output should be (vocab_size,) for single token prediction
        prediction = self.fc(out_lstm.squeeze(0).squeeze(0))
        return prediction, hidden, cell

class Seq2Seq(nn.Module):
    def __init__(self, encoder, decoder):
        super().__init__()
        self.encoder = encoder
        self.decoder = decoder

    def forward(self, src, trg):
        hidden, cell = self.encoder(src)
        input = trg[0]
        outputs = []
        for t in range(1, len(trg)):
            out, hidden, cell = self.decoder(input, hidden, cell)
            outputs.append(out)
            input = trg[t]
        # Stack outputs. Each 'out' is now (vocab_size,), so stack creates (seq_len, vocab_size)
        return torch.stack(outputs)

# -------------------- TRAIN --------------------
VOCAB_SIZE = len(vocab)
encoder = Encoder(VOCAB_SIZE, 16, 32)
decoder = Decoder(VOCAB_SIZE, 16, 32)
model = Seq2Seq(encoder, decoder)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.05)

for _ in range(300):
    optimizer.zero_grad()
    output = model(src_tensor, trg_tensor)
    # output from model is now (seq_len, vocab_size), trg_tensor[1:] is (seq_len,)
    loss = criterion(output, trg_tensor[1:])
    loss.backward()
    optimizer.step()

# -------------------- INFERENCE --------------------
def summarize(sentence):
    src = [vocab["<sos>"]] + encode(sentence) + [vocab["<eos>"]]
    src_tensor = torch.tensor(src).unsqueeze(1)

    hidden, cell = encoder(src_tensor)
    input = torch.tensor(vocab["<sos>"])
    result = []

    for _ in range(5):
        out, hidden, cell = decoder(input, hidden, cell)
        # 'out' is now (vocab_size,)
        word_id = out.argmax(0).item()
        if word_id == vocab["<eos>"]:
            break
        result.append(inv_vocab[word_id])
        input = torch.tensor(word_id)

    return f"[{'; '.join(result)}]"

print(summarize("apple is a healthy fruit here"))
```
